```python
from langchain_community.document_loaders import UnstructuredWordDocumentLoader

# Install pi_heif as it's a dependency for unstructured to process certain image formats in PDFs
!pip install pi_heif -q
!pip install unstructured-inference -q
!pip install pdf2image -q
!pip install unstructured-pytesseract -q
!apt-get install poppler-utils -y -q

# Install python-docx to handle .docx files
!pip install python-docx -q

loader = UnstructuredWordDocumentLoader("/content/Customer Support Knowledge Base.docx")
documents = loader.load()

print("Pages:", len(documents))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = text_splitter.split_documents(documents)
print("Chunks:", len(chunks))
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2"
)
```

```python
from langchain_community.vectorstores import Chroma

vectorstore = Chroma.from_documents(
    chunks,
    embedding_model
)

retriever = vectorstore.as_retriever(search_kwargs={"k": 3})
```

```python
from transformers import pipeline

llm_pipeline = pipeline(
    "text-generation",
    model="google/flan-t5-base"
)

def local_llm(prompt):
    result = llm_pipeline(prompt, max_length=200)
    return result[0]["generated_text"]
```

```python
def process_query(state):
    query = state["query"]

    # Retrieve relevant chunks
    docs = retriever.invoke(query)

    if not docs:
        return {
            "query": query,
            "response": "No relevant information found.",
            "confidence": 0.2
        }

    context = "\n".join([doc.page_content for doc in docs])

    prompt = f"""
    You are a customer support assistant.
    Answer ONLY using the context below.

    Context:
    {context}

    Question:
    {query}
    """

    answer = local_llm(prompt)

    # Confidence logic
    confidence = 0.9 if len(docs) > 0 else 0.3

    return {
        "query": query,
        "response": answer,
        "confidence": confidence
    }
```

```python
def route_decision(state):
    if state["confidence"] < 0.6:
        return "escalate"
    return "answer"
```

```python
def human_node(state):
    print("\n⚠️ Escalation Triggered")
    print("Query:", state["query"])

    human_answer = input("Enter human response: ")

    return {
        "response": human_answer,
        "confidence": 1.0,
        "escalated": True
    }
```

```python
def output_node(state):
    print("\n✅ Final Answer:")
    print(state["response"])
    return state
```

```python
from langgraph.graph import StateGraph

class GraphState(dict):
    pass

builder = StateGraph(GraphState)

# Nodes
builder.add_node("process", process_query)
builder.add_node("hitl", human_node)
builder.add_node("output", output_node)

# Entry
builder.set_entry_point("process")

# Conditional routing
builder.add_conditional_edges(
    "process",
    route_decision,
    {
        "answer": "output",
        "escalate": "hitl"
    }
)

# HITL → Output
builder.add_edge("hitl", "output")

graph = builder.compile()
```

```python
from typing import TypedDict

class GraphState(TypedDict, total=False):
    query: str
    response: str
    confidence: float
    escalated: bool
```

```python
def process_query(state):
    query = state.get("query", "")

    if not query:
        return {
            "response": "Invalid query",
            "confidence": 0.0
        }

    docs = retriever.invoke(query)

    if not docs:
        return {
            "query": query,
            "response": "No relevant information found.",
            "confidence": 0.2
        }

    context = "\n".join([doc.page_content for doc in docs])

    prompt = f"""
    Answer using ONLY the context below.

    Context:
    {context}

    Question:
    {query}
    """

    answer = local_llm(prompt)

    return {
        "query": query,
        "response": answer,
        "confidence": 0.9
    }
```

```python
def route_decision(state):
    confidence = state.get("confidence", 0)

    if confidence < 0.6:
        return "escalate"
    return "answer"
```

```python
def human_node(state):
    print("\n⚠️ Escalation Triggered")
    print("Query:", state.get("query", ""))

    human_answer = input("Enter human response: ")

    return {
        "query": state.get("query", ""),
        "response": human_answer,
        "confidence": 1.0,
        "escalated": True
    }
```

```python
def output_node(state):
    print("\n✅ Final Answer:")
    print(state.get("response", "No response"))
    return state
```

```python
from langgraph.graph import StateGraph

builder = StateGraph(GraphState)

builder.add_node("process", process_query)
builder.add_node("hitl", human_node)
builder.add_node("output", output_node)

builder.set_entry_point("process")

builder.add_conditional_edges(
    "process",
    route_decision,
    {
        "answer": "output",
        "escalate": "hitl"
    }
)

builder.add_edge("hitl", "output")

graph = builder.compile()
```

```python
while True:
    query = input("\nAsk your question (type 'exit'): ")

    if query.lower() == "exit":
        break

    graph.invoke({
        "query": query,
        "response": "",
        "confidence": 0.0
    })
```
